## Data analyse
Per tijdstap, wij hebben 20 seconden gekozen, meten wij de temperatuur. Hierna berekenen wij met de totale gepasseerde tijd en de totale verandering in temperatuur de soortelijke warmte van water.

# Resultaten


````python
# Hier de data en de analyse
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
m = 0.300 
U = 17.2
I = 1.78
# 17.2 V, 1.78 A
t = np.array([0,20,40,60,80,100,120,140,160,180,200])
T = np.array([16.7,17.6,18.2,18.7,19.2,19.7,20.1,20.6,21.1,21.5,22])
def function(t,a,b):
    return a*t+b
values, covariance = curve_fit(function,t,T)
x_fit = np.linspace(1,200,1000)
y_fit = function(x_fit,*values)
plt.figure()
plt.plot(t,T,'k.')
plt.plot(x_fit,y_fit,'r--')
plt.xlabel(r't(s)')
plt.ylabel(r'T(°C)')
plt.show()
````

````python
P = I*U
print(P)
Q =P*t[-1]
dt = T[-1]-T[0]
cw = Q/(m*dt)
print(f'De soortelijke warmte van water is {cw:.2f} J/(kgK)')
````

# Discussie en conclusie
In dit experiment komen wij uit op een soortelijke warmte voor water van 3851.07 J/(kgK). De officiele waarde voor de soortelijke warmte van water is gelijk aan 4186 J/(kgK). 
De afwijking van de echte waarde die wij hebben komt zeer waarschijnlijk door warmteverliezen aan omgeving en lucht, ook vooral omdat er geen dop aanwezig was op deze maatbekers. Wanneer de maatbeker beter geïsoleerd is, zal deze waarde voor de soortelijke warmte accurater worden.
